1.12. Crear la función para automatizar todo el proceso, todos los procesos anteriores que hemos ido realizando los encapsularemos dentro de una función. La función se llamará ``transform_data()``, el único requisito es que tenga el parámetro de la ruta donde se encuentran los datos. Utilizar la # para nombrar lo que se hace en cada parte del código.



```python
def transform_data(datapath):
    #leemos los datos desde el archivo
    liga21 = pd.read_csv(datapath)
    #Sustituimos los valores NAN por 0 y guardamos el cambio.
    liga21.fillna(0, inplace=True)
    #Eliminamos la columna 'Rk' y la columna 'Matches'
    liga21.drop(['Rk', 'Matches'], axis=1, inplace=True)
    #Solucionamos el problema de los nombres de los jugadores
    liga21['Player']= liga21.Player.apply(lambda name: name.split('\\')[0])
    #Solucionamos el problema de la nacionalidad
    liga21['Nation']= liga21.Nation.apply(lambda nation: nation.split(' ')[1])
    #Solucionamos el problema de las posiciones, quedandonos sólo con la primera
    liga21['Pos']= liga21.Pos.apply(lambda pos : pos[:2] if len (pos) > 2 else pos)
    #Cambiamos los nombres de las columnas
    liga21 = liga21.rename(columns ={
    'Gls.1':'Gls/90',
    'Ast.1':'Ast/90',
    'G-PK.1':'G-PK/90',
    'xG.1':'xG/90',
    'xA.1':'xA/90',
    'npxG.1':'npxG/90',
    'npxG+xA.1':'npxG+xA/90'
    })
    return liga21
```

Comprobar que nuestra función funciona correctamente, utilizando el path en donde se encuentran los datos. Guardar el resultado de salida en una variable y aplicarle ``.head()`` para comprobar que se han realizado todos las transformaciones.

```python
test_function = transform_data('fbref_laliga21.csv')
test_function.head()
```

```python
liga21.Gls.unique()
```

```python
for i in range(len(liga21)):
  if liga21.Gls.values[i] <=15:
    liga21.loc[i, 'type_goal_player'] = 'Poco Goleador'

  elif (liga21.Gls.values[i] > 15) & (liga21.Gls.values[i] <=20):
    liga21.loc[i, 'type_goal_player'] = 'Goleador'

  else:
    liga21.loc[i, 'type_goal_player'] = 'Muy goleador'
```
